```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import statsmodels.api as sm 
data = pd.read_csv('1.01.simple linear regression.csv.csv')
data
```

```python
data.describe()
```

```python
y = data['GPA']
x1 = data['SAT']
```

```python
plt.scatter(x1,y)
plt.xlabel('SAT',fontsize=20)
plt.ylabel('GPA',fontsize=20)
plt.show() 
```

```python
x = sm.add_constant(x1)
results = sm.OLS(y,x).fit()
results.summary()
```

```python
plt.scatter(x1,y)
yhat = 0.0031*x1 + 0.275
data = plt.plot(x1,yhat,lw=4, c='orange', label='regression line')
plt.xlabel('SAT',fontsize = 10)
plt.ylabel('GPA', fontsize = 10)
plt.show()
```
